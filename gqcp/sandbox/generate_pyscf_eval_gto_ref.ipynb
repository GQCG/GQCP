```python
import pyscf
import numpy as np
```

```python
mol= pyscf.gto.Mole()
mol.build(
    atom = "H 0.0 0.0 0.0; O 1.0 0.0 0.0; H 2.0 0.0 0.0",
    basis='sto-3g',
    unit='bohr',
)

ao_vals = mol.eval_gto('GTOval_cart', np.array([[1.3, -0.9, 3.7]]))

for val in ao_vals[0]:
    print(np.around(val, 9))
```
