```python
import pyscf
import numpy as np
```

```python
mol= pyscf.gto.Mole()
mol.build(
    atom = "H 0.0 0.0 0.0; O 1.0 0.0 0.0; H 2.0 0.0 0.0",
    basis='sto-3g',
    unit='bohr',
)

ao_vals = mol.eval_gto('GTOval_cart', np.array([[1.3, -0.9, 3.7]]))

for val in ao_vals[0]:
    print(np.around(val, 9))
```

```python
grad_ao_vals = mol.eval_gto('GTOval_ip_cart', np.array([[1.3, -0.9, 3.7]])) # Nabla AO

grad_ao_vals_x, grad_ao_vals_y, grad_ao_vals_z = grad_ao_vals
```

```python
for val in grad_ao_vals_x[0]:
    print(np.around(val, 9))
```

```python
for val in grad_ao_vals_y[0]:
    print(np.around(val, 9))
```

```python
for val in grad_ao_vals_z[0]:
    print(np.around(val, 9))
```
